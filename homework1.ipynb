Q7. Sum of weights

    Select all the cars from Asia
    Select only columns vehicle_weight and model_year
    Select the first 7 values
    Get the underlying NumPy array. Let's call it X.
    Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
    Invert XTX.
    Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
    Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
    What's the sum of all the elements of the result?

    Note: You just implemented linear regression. We'll talk about it in the next lesson.

    0.051
    0.51
    5.1
    51


```python
#From Q5, variable Asia_origin contains all vehicles from Asia
Asia_df=Asia_origin[["vehicle_weight","model_year"]]
seven_val_df=Asia_df[:7]
seven_val_df.shape

#underlying numpy array
X=seven_val_df.to_numpy()
print("Matrix X: ",X)
print("---------------------------------------------------")

#transposing the matrix
transposed_X=X.T
print("X Transpose: ",transposed_X)
print("---------------------------------------------------")

#matrix multiplication X^T * X
XTX=np.matmul(transposed_X,X)
print("X^T * X: ",XTX)
print("---------------------------------------------------")

#calculating inverse of X^T * X
#first check the determinant to see if the matrix is invertible
det=np.linalg.det(XTX)
print("Determinant: ",det)
print("---------------------------------------------------")
if det!=0:
    XTX_inv=np.linalg.inv(XTX)
    print("Inverse of X^T * X: ",XTX_inv)
    print("---------------------------------------------------")

#create y array
list=[1100, 1300, 800, 900, 1000, 1100, 1200]
y=np.array(list)
print("y: ",y)
print("---------------------------------------------------")

#Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. 
#Call the result w.
w=np.matmul(XTX_inv,transposed_X)
w=np.matmul(w,y)
print("w: ",w)

#sum of w elements
print("Sum of w elements: ",w.sum())
```
